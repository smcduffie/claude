### Part (b) — Express force as a percentage of body weight

Force in Newtons depends on how heavy the subject is. To compare subjects you first divide by body weight.

`SiCo01` has a mass of 56 kg. Convert that to a weight in Newtons (mass times 9.81), then use **broadcasting** to turn `total_left` and `total_right` into percentages of body weight. Report the peak value of each foot.

````python
mass_kg = 56.0
body_weight_N = mass_kg * 9.81
print(f'body weight: {body_weight_N:.1f} N')

# rememebrt that one number divides a whole array: that is broadcasting

# Write your code here
total_left_bw  = total_left  / body_weight_N * 100
total_right_bw = total_right / body_weight_N * 100

print(f'peak left foot:  {total_left_bw.max():.1f} % of body weight')
print(f'peak right foot: {total_right_bw.max():.1f} % of body weight')
````

### Part (c) — Find when each foot is on the floor

A foot is on the floor when the total force under it is above 20 N.

Use the boolean mask provided for each foot (`loaded_left` and `loaded_right`) and report the fraction of the recording in which each foot is loaded. **Hint:** the mean of a boolean array is the fraction of `True` values.

In the answer cell: the two fractions add up to more than 1. Explain how that is possible.

````python
loaded_left  = total_left  > 20
loaded_right = total_right > 20
#note that you need to define the variable total_left and total_right before using them in the above code.

# Write your code here
frac_left  = loaded_left.mean()
frac_right = loaded_right.mean()
print(f'fraction of the recording with the left foot loaded:  {frac_left:.3f}')
print(f'fraction of the recording with the right foot loaded: {frac_right:.3f}')
print(f'sum of the two fractions: {frac_left + frac_right:.3f}')

# time when both feet are on the floor at the same time (double support)
both = loaded_left & loaded_right
print(f'fraction with both feet loaded: {both.mean():.3f}')
````

**Your answer:** The left foot is loaded 0.659 of the time and the right foot 0.651, which adds up to 1.31. That is possible because the fractions overlap: in walking there are periods of **double support**, when both feet are on the floor at the same time (at the transition from one step to the next). Both feet are loaded about 0.31 of the recording, and that time is counted in both fractions. Only a gait with no double support, such as running, would give a sum of 1 or less.

### Part (a) — Wrap your work in a function

The following function takes a file name and a body mass, and returns:

- the peak force under each foot, in percentage of body weight
- the fraction of the recording in which each foot is loaded
- the mean and the standard deviation of the per-second force under the left foot, in percentage of body weight

Execute the following cell to ensure the function exists in the notebook memory

````python
def summarize(file_name, mass_kg):
    """Summarize one force recording. Returns a dictionary."""
    data = np.loadtxt(file_name)
    tot_left, tot_right = data[:, 17], data[:, 18]

    body_weight_N = mass_kg * 9.81
    left_bw  = tot_left  / body_weight_N * 100
    right_bw = tot_right / body_weight_N * 100

    n = (len(left_bw) // 100) * 100
    per_second = left_bw[:n].reshape(-1, 100).mean(axis=1)

    return {
        'peak_left_bw'   : left_bw.max(),
        'peak_right_bw'  : right_bw.max(),
        'loaded_left'    : (tot_left  > 20).mean(),
        'loaded_right'   : (tot_right > 20).mean(),
        'per_second_mean': per_second.mean(),
        'per_second_std' : per_second.std(),
        'per_second'     : per_second,
    }
````

Read the function and get a complete understanding of what it is doing.


where `XXX` are the results obtained for each number.

### Part (c) — Compare the two groups

Group the table by `group` and report the mean of `avg_speed`, `cadence`, `avg_step_length` and `cv_stride_time_left`.

Then check this identity on the two group means:

$$ \text{speed (m/s)} = \frac{\text{cadence (steps/min)} \times \text{step length (m)}}{60} $$

confirm that this measure of `speed` correspond to the reported `avg_speed`.

In the answer cell: the two groups walk at clearly different speeds but at almost the same cadence. Cadence is steps per minute. Using the identity above, what does that tell you about **how** the slower group is slower?

````python
group_means = gait.groupby('group')[['avg_speed', 'cadence', 'avg_step_length', 'cv_stride_time_left']].mean()
display(group_means)

# check the identity: speed = cadence * step length / 60
group_means['speed_from_identity'] = group_means['cadence'] * group_means['avg_step_length'] / 60
display(group_means[['avg_speed', 'speed_from_identity']])
````

**Your answer:** The group means are: control 1.24 m/s, 109.2 steps/min, 0.69 m step length; PD 1.02 m/s, 108.8 steps/min, 0.57 m step length. The identity gives 109.2 × 0.688 / 60 = 1.25 m/s for controls and 108.8 × 0.565 / 60 = 1.03 m/s for PD, which matches the reported `avg_speed` (1.24 and 1.02 m/s). The small gap comes from applying the identity to group means (the mean of a product is not exactly the product of the means) and from the missing values in `avg_speed` and `avg_step_length`.

Speed = cadence × step length / 60, and cadence is almost the same in both groups (about 109 steps/min). So the slower speed of the PD group comes almost entirely from **shorter steps** (0.57 m vs 0.69 m, about 18 % shorter), not from taking fewer steps per minute. The PD group takes steps just as often but covers less ground with each one, which fits the short, shuffling steps typical of Parkinsonian gait. The PD group also has a higher `cv_stride_time_left` (0.22 vs 0.15), meaning their stride timing is more variable.
